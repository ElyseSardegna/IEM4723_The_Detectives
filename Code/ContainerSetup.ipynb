```python
import gurobipy as gp
from gurobipy import GRB

m = gp.Model("test")
x = m.addVar(name="x")
y = m.addVar(name="y")
m.setObjective(x + y, GRB.MAXIMIZE)
m.addConstr(x + 2 * y <= 4)
m.addConstr(4 * x + 3 * y <= 12)
m.optimize()

print(m.objVal, x.X, y.X)
```
